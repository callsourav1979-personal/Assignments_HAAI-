```python
import sys
import torch
import transformers
from transformers import T5Tokenizer, T5ForConditionalGeneration
import re

#####
transformers.logging.set_verbosity_error()
transformers.utils.logging.disable_progress_bar()
```

```python

def llm_function(model,tokenizer,context,question):

  # Generate a deterministic output either 'YES' or 'NO' answering the question using the provided context
  q = context + " " + question

  prompt = f"""
  You are a knowledgable assistant. Only respond in YES or NO. No additional information is required.
  Question:
     {q}

    Answer:
    """
 # Tokenize Prompt (Encoder Input)
  inputs = tokenizer(prompt, return_tensors="pt")
  input_ids = inputs["input_ids"]
  attention_mask = inputs["attention_mask"]

 # Decoder start token (<pad> for T5)
  decoder_input_ids = torch.tensor([[model.config.decoder_start_token_id]])

  with torch.no_grad():
         outputs = model(
                    input_ids=input_ids,
                    attention_mask=attention_mask,
                    decoder_input_ids=decoder_input_ids
                )
         logits = outputs.logits[0,0]

         #Get the token IDs for YES and NO
         yes_token_id = tokenizer("YES",add_special_tokens=False).input_ids[0]
         no_token_id = tokenizer("NO",add_special_tokens=False).input_ids[0]

         #Extract logits
         yes_logit = logits[yes_token_id].item()
         no_logit = logits[no_token_id].item()

         #select the answer with highest logits
         if yes_logit > no_logit:
            final_output = "YES"
         else:
            final_output = "NO"

         return final_output
```

    1. Engineer the prompt using the query and the context.
    2. Tokenize the prompt.
    3. Generate output for the prompt.
    4. Extract the logit values to determine the output
    5. Format the output to be exactly YES or NO.
